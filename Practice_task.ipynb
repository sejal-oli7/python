##TO-Do LIst

##Organizing Student Grades

```python
# Store student grades
grades = [85, 90, 78, 88, 95]

# Calculate average
average = sum(grades) / len(grades)

print("Grades:", grades)
print("Average Grade:", average)
```

##Users Feedback

```python
# Collect user feedback
feedback = ["Good service", "Fast delivery", "Excellent quality"]

# Add more feedback
feedback.append("Affordable price")
feedback.append("Friendly staff")

# Display feedback
print("User Feedback:")
for item in feedback:
    print(item)

# Total feedback
print("Total Feedback:", len(feedback))
```

##Calculate Sum and Average

##Count Even and Odd Numbers

```python
numbers = [10, 21, 4, 45, 66, 93, 11]

even = 0
odd = 0

for num in numbers:
    if num % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```
