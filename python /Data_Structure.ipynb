# Data Structure in Python


#   Introduction to Lists

## What is a List?

A **list** is a collection of multiple items stored in a single variable.

**Properties of Lists**
- Ordered
- Mutable (can be changed)
- Allows duplicate values
- Can store different data types

---

# 1. Creating a List

```python
fruits = ["Apple", "Banana", "Mango"]
numbers = [10, 20, 30, 40]

print(fruits)
print(numbers)
```

**Output**

```
['Apple', 'Banana', 'Mango']
[10, 20, 30, 40]
```

---

# 2. Access a List Element

Lists use **indexing**.

| Index | Value |
|------|------|
|0|Apple|
|1|Banana|
|2|Mango|

```python
fruits = ["Apple", "Banana", "Mango"]

print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[-1])
```

**Output**

```
Apple
Banana
Mango
Mango
```

---

# 3. Modify a List Element

Lists are mutable, so values can be changed.

```python
fruits = ["Apple", "Banana", "Mango"]

fruits[1] = "Orange"

print(fruits)
```

**Output**

```
['Apple', 'Orange', 'Mango']
```

---

# 4. List Methods

## append()

Adds an item at the end.

```python
numbers = [1,2,3]

numbers.append(4)

print(numbers)
```

Output

```
[1,2,3,4]
```

---

## insert()

Adds an item at a specific index.

```python
numbers = [1,2,3]

numbers.insert(1,100)

print(numbers)
```

Output

```
[1,100,2,3]
```

---

## remove()

Removes an item by value.

```python
numbers = [1,100,2,3]

numbers.remove(100)

print(numbers)
```

Output

```
[1,2,3]
```

---

## pop()

Removes an item by index.

```python
numbers = [1,2,3]

numbers.pop(1)

print(numbers)
```

Output

```
[1,3]
```

---

## sort()

Sorts the list in ascending order.

```python
numbers = [5,3,1,4,2]

numbers.sort()

print(numbers)
```

Output

```
[1,2,3,4,5]
```

---

## reverse()

Reverses the list.

```python
numbers = [1,2,3]

numbers.reverse()

print(numbers)
```

Output

```
[3,2,1]
```

---

## count()

Counts how many times a value appears.

```python
numbers = [1,2,1,3,1]

print(numbers.count(1))
```

Output

```
3
```

---

## index()

Returns the index of a value.

```python
numbers = [10,20,30]

print(numbers.index(20))
```

Output

```
1
```

---

## extend()

Adds another list.

```python
a = [1,2]
b = [3,4]

a.extend(b)

print(a)
```

Output

```
[1,2,3,4]
```

---

## clear()

Removes all elements.

```python
numbers = [1,2,3]

numbers.clear()

print(numbers)
```

Output

```
[]
```

---

# 5. Slicing

Syntax

```python
list[start:stop:step]
```

```python
numbers = [10,20,30,40,50]

print(numbers[:3])
print(numbers[2:])
print(numbers[1:4])
print(numbers[::2])
print(numbers[::-1])
```

Output

```
[10,20,30]
[30,40,50]
[20,30,40]
[10,30,50]
[50,40,30,20,10]
```

# 6. Iteration

Using a **for loop**

```python
fruits = ["Apple","Banana","Mango"]

for fruit in fruits:
    print(fruit)
```

Output

```
Apple
Banana
Mango
```

Using **range()**

```python
fruits = ["Apple","Banana","Mango"]

for i in range(len(fruits)):
    print(fruits[i])
```

Using **while loop**

```python
fruits = ["Apple","Banana","Mango"]

i = 0

while i < len(fruits):
    print(fruits[i])
    i += 1
```

# 7. List Comprehension (Very Important)

List comprehension creates a new list in a single line.

Syntax

```python
[new_item for item in iterable]
```

Example 1

```python
numbers = [1,2,3,4,5]

square = [x*x for x in numbers]

print(square)
```

Output

```
[1,4,9,16,25]
```

Example 2 (Even Numbers)

```python
numbers = [1,2,3,4,5,6]

even = [x for x in numbers if x % 2 == 0]

print(even)
```

Output

```
[2,4,6]
```

Example 3 (Uppercase)

```python
names = ["ram","hari","sita"]

upper = [name.upper() for name in names]

print(upper)
```

Output

```
['RAM', 'HARI', 'SITA']
```

# 8. Nested Lists

A nested list is a list inside another list.

```python
matrix = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]

print(matrix)
```

Access an element

```python
print(matrix[0][1])
```

Output

```
2
```

Modify an element

```python
matrix[1][2] = 100

print(matrix)
```

Output

```
[[1,2,3],[4,5,100],[7,8,9]]
```

Loop through a nested list

```python
for row in matrix:
    for item in row:
        print(item)
```

Output

```
1
2
3
4
5
6
7
8
9
```
