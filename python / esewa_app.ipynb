```python
import random
```

```python
class DigitalWallet:

    def __init__(self, owner_name, wallet_id, balance=0):
        # Yo wallet ko owner ko naam ho
        self.owner_name = owner_name

        # Harek wallet ko unique ID huncha
        self.wallet_id = wallet_id

        # Balance lai private rakheko cha
        # Bahira bata directly change garna nadina ko lagi
        # yo Encapsulation ho
        self.__balance = balance

        # Normal wallet ko maximum limit 50,000 ho
        self.max_limit = 50000

        # Sabai transaction ko record rakhna list banako
        self.transaction_history = []

    # Balance herna ko lagi method
    # Direct __balance access nagari yo method use garincha
    def get_balance(self):
        return self.__balance

    # Wallet ma balance add garna ko lagi
    def add_balance(self, amount):
        self.__balance += amount

    # Wallet bata balance ghatauna ko lagi
    def deduct_balance(self, amount):
        self.__balance -= amount


#Load money
    def load_money(self, amount):

        # Amount 0 wa negative bhayo bhane accept gardaina
        if amount <= 0:
            print(" Amount 0 bhanda thulo hunu parcha")
            return

        # Balance ra load garna lageko amount
        # maximum limit bhanda badi hunu hudaina
        if self.__balance + amount > self.max_limit:
            print(" Wallet ko maximum limit exceed bhayo")
            return

        # Balance ma amount add garne
        self.__balance += amount

        # Transaction history ma record rakhne
        self.transaction_history.append(
            f"Loaded Rs. {amount} | Success"
        )

        print(f" Rs. {amount} successfully load bhayo")
        print(f"Current Balance: Rs. {self.__balance}")

  #pay bill

    def pay_bill(self, amount, description):

        # Amount valid cha ki chaina check garne
        if amount <= 0:
            print(" Amount 0 bhanda thulo hunu parcha")
            return

        # Wallet ma bill tirna pugne paisa cha ki chaina check garne
        if amount > self.__balance:
            print(" Payment Failed: Balance pugena")

            # Failed transaction pani history ma rakhne
            self.transaction_history.append(
                f"{description} | Rs. {amount} | Failed"
            )
            return

        # Bill pay bhayepachi balance bata amount ghataune
        self.__balance -= amount

        # Successful transaction history ma save garne
        self.transaction_history.append(
            f"{description} | Rs. {amount} | Success"
        )

        print(f" {description} successfully paid")
        print(f"Paid Amount: Rs. {amount}")
        print(f"Remaining Balance: Rs. {self.__balance}")


#transfer money
    def transfer_money(self, receiver, amount):

        # Amount valid cha ki chaina check garne
        if amount <= 0:
            print(" Amount 0 bhanda thulo hunu parcha")
            return

        # Afnai wallet ma transfer garna nadine
        if receiver == self:
            print(" Afailai paisa transfer garna mildaina")
            return

        # Sender ko wallet ma enough balance cha ki chaina check garne
        if amount > self.__balance:
            print(" Transfer Failed: Balance pugena")
            return

        # Sender ko balance bata paisa ghataune
        self.__balance -= amount

        # Receiver ko balance ma paisa add garne
        receiver.__balance += amount

        # Sender ko transaction history update garne
        self.transaction_history.append(
            f"Sent Rs. {amount} to {receiver.owner_name} | Success"
        )

        # Receiver ko transaction history pani update garne
        receiver.transaction_history.append(
            f"Received Rs. {amount} from {self.owner_name} | Success"
        )

        print(
            f" Rs. {amount} {receiver.owner_name} lai successfully transfer bhayo"
        )


#Check status
    def check_status(self):

        print("\n========== WALLET STATUS ==========")

        # Wallet ko basic information dekhaune
        print(f"Owner      : {self.owner_name}")
        print(f"Wallet ID  : {self.wallet_id}")
        print(f"Balance    : Rs. {self.__balance}")
        print(f"Max Limit  : Rs. {self.max_limit}")

        print("\n----- Transaction History -----")

        # History empty cha bhane message dekhaune
        if len(self.transaction_history) == 0:
            print("Ahile samma kunai transaction chaina.")

        # History ko harek transaction print garne
        else:
            for transaction in self.transaction_history:
                print("-", transaction)

        print("====================================")



# PREMIUM WALLET
class PremiumWallet(DigitalWallet):

    def __init__(self, owner_name, wallet_id, balance=0):

        # Parent class DigitalWallet ko constructor call gareko
        # Yasle parent ko attributes initialize garcha
        super().__init__(
            owner_name,
            wallet_id,
            balance
        )

        # Premium wallet ko limit normal bhanda dherai cha
        self.max_limit = 500000

        # Premium wallet ma 2% reward/cashback cha
        self.cash_rate = 0.02


    # PREMIUM PAY BILL

    def pay_bill(self, amount, description):

        # Amount valid cha ki chaina check garne
        if amount <= 0:
            print(" Amount 0 bhanda thulo hunu parcha")
            return

        # Bill tirna pugne balance cha ki chaina check garne
        if amount > self.get_balance():
            print(" Payment Failed: Balance pugena")

            self.transaction_history.append(
                f"{description} | Rs. {amount} | Failed"
            )
            return

        # Bill pay garepachi balance bata paisa ghataune
        self.deduct_balance(amount)

        # Premium wallet ko 2% reward calculate garne
        reward = amount * self.cash_rate

        # Reward feri wallet ko balance ma add garne
        self.add_balance(reward)

        # Transaction history ma payment ra reward dubai rakhne
        self.transaction_history.append(
            f"{description} | Paid: Rs. {amount} | "
            f"Reward: Rs. {reward} | Success"
        )

        print(f"{description} successfully paid")
        print(f"Paid Amount : Rs. {amount}")
        print(f"Reward      : Rs. {reward}")
        print(f"Final Balance: Rs. {self.get_balance()}")



# NORMAL WALLET KO TEST

# Sejal ko wallet create gareko
# Starting balance = Rs. 5,000
sejal = DigitalWallet(
    "Sejal",
    "W001",
    5000
)

# Asmi ko wallet create gareko
# Starting balance = Rs. 10,000
asmi = DigitalWallet(
    "Asmi",
    "W002",
    10000
)


print("\n========== NORMAL WALLET ==========")

# Sejal le Rs. 5,000 load garcha
# 5,000 + 5,000 = 10,000
sejal.load_money(5000)

# Sejal le Rs. 2,000 ko electricity bill tircha
# 10,000 - 2,000 = 8,000
sejal.pay_bill(
    2000,
    "Electricity Bill"
)

# Sejal le asmi lai Rs. 1,000 transfer garcha
# Sejal = 8,000 - 1,000 = 7,000
# Asmi = 10,000 + 1,000 = 11,000
sejal.transfer_money(
    asmi,
    1000
)

# Sejal ko wallet status herne
sejal.check_status()

# Ashish ko wallet status herne
asmi.check_status()


# PREMIUM WALLET KO TEST
# Premium wallet create gareko
# Starting balance = Rs. 5,000
premium = PremiumWallet(
    "Sejal",
    "P001",
    5000
)

print("\n========== PREMIUM WALLET ==========")

# Premium wallet bata Rs. 2,000 ko bill pay
premium.pay_bill(
    2000,
    "Internet Bill"
)

# Premium wallet ko status herne
premium.check_status()
```
