```python
#@ try, execpt and else block

try:
  num = int(input("Enter a number : "))
  result = 10/num
except ZeroDivisionError as ex:
  print(ex)
  print("Denominator should always be greater than 0")
except NameError as ex:
  print(ex)
  print("I am NameError")
else:
  print("No execption")
  print(result)
```

```python
#@ try, execpt , else and finally

try:
  num = int(input("Enter a number : "))
  result = 10/num
except ZeroDivisionError as ex:
  print(ex)
  print("Denominator should always be greater than 0")
except NameError as ex:
  print(ex)
  print("I am NameError")
else:
  print("No execption")
  print(result)
finally:
  print("I will always execute")
```
