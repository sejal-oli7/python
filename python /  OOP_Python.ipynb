## The self Parameter

````python
class Person:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def greet(self):
    print("Hello, my name is " + self.name)

# Create an object of the Person class

person1 = Person("Sejal", 36)

person1.greet()
````

## Python Class Methods

. Methods are functions that belongs to a class.


. It defines the behaviors of a object created from the class

````python
class Person:
  def __init__(self, name):
    self.name = name

  def greet(self):
    print("Hello sir, I am" + self.name)
p1 = Person("Sejal")
p1.greet()
````

````python
#calculator
class Calculator:
    def add(self, a, b):
        return a + b

    def multiple(self, a, b):
        return a * b

    def divide(self, a, b):
        return a / b

    def subtraction(self, a, b):
        return a - b

c = Calculator()            # Calculator ko object banayo

print(c.add(10, 5))         # add method call garyo
print(c.multiple(10, 5))    # multiple method call garyo
print(c.divide(10, 5))      # divide method call garyo
print(c.subtraction(10, 5)) # subtraction method call garyo
````

##Multiple Methods

. A class can have multiple methods that work together

````python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []          # songs ko empty list

    def add_song(self, song):
        self.songs.append(song)  # song list ma add garxa
        print(f"Added a song: {song}")

    def remove_song(self, song):
        if song in self.songs:
            self.songs.remove(song)  # song remove garxa
            print(f"Removed a song: {song}")

    def show_songs(self):
        print(f"Playlist: {self.name}")  # playlist ko name print garxa
        for song in self.songs:
            print(song)                  # harek song print garxa


my_playlist = Playlist("Favourites")  # object banayo

my_playlist.add_song("Song 1")
my_playlist.add_song("Song 2")
my_playlist.add_song("Song 3")

my_playlist.show_songs()              # songs dekhaune
````

````python
my_playlist.remove_song("Song 3")
my_playlist.show_songs()
````

##Inheritance in Python
 Inheritance  allows use to define a class that iherits all the methods and properties from another class

 - *** parent class*** : Class being inherited from, also calll it as ```base class```

 - ***Child class*** : Class that inherits from another class, also call it as derived class

````python
#@ Create a Parent Class
## Any class can be a parent class, so that the syntax remains same as above class defination
````

````python
class Person:
    def __init__(self, fname, lname):
        self.firstname = fname
        self.lastname = lname

    def printname(self):
        print(self.firstname, self.lastname)


# Person ko object banayo
person1 = Person("Sejal", "Oli")

# printname method call garyo
person1.printname()
````
