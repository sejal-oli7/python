````python
class Mother : # parent class 1
  mothername = ""

  def mother(self):
    print(self.mothername)

class Father: # parent class 2
  fathername = ""

  def father(self):
    print(self.fathername)

class Son(Mother, Father): # child class
  def parents(self):
    print("Father :", self.fathername)
    print("Mother :", self.mothername)

s1 = Son()
s1.fathername = "RAM"
s1.mothername = "Sarmila"
s1.parents()
````

## Multilevel Inheritance


````python
#MUltilevel Inheritance grandfather and child

class Grandfather:
  def __init__(self, grandfathername):
    self.grandfathername = grandfathername

class Father(Grandfather):
  def __init__(self, fathername, grandfathername):
    self.fathername = fathername
    Grandfather.__init__(self, grandfathername)

class Son(Father):
  def __init__(self, sonname, fathername, grandfathername):
    self.sonname = sonname
    Father.__init__(self, fathername, grandfathername)

  def print_name(self):
    print('Grandfather name :', self.grandfathername)
    print("Father name :", self.fathername)
    print("Son name :", self.sonname)

s1 = Son('Simon', 'Ram prasad', 'Krishna prasad')
print(s1.grandfathername)
s1.print_name()
````

## Important Notes:

. Son inherits from Father, and Father inherits from GRandFather

. Each constructor passes values up the inheritance chain using explicit constructor calls

. All ancestors class attributes are accessible from the bottom most class which is Son is our case

##Super() function

. super()function that will make the child class inherits all the methods and properties from it's parent class

. It is commonly used in the child class's init() method to initalize inherited attributes.

````python
#@ Animal
#Parent Class
class Animal:
    def __init__(self, name):
        self.name = name

    def info(self):
        print("Animal name:", self.name)


# Child Class
class Dog(Animal):
    def __init__(self, name, breed):
        # Call parent class constructor
        super().__init__(name)
        self.breed = breed

    def details(self):
        print(self.name, "is a", self.breed)


d = Dog("Max", "Husky")

d.info()
d.details()
````

## Takeway:

. ```super()``` function is used ```__init__()``` methof of Dog to call the constructor of ```Animal``` and initialize inherited attribute(name)

. It ensures that parent class functionality is reused without needing to rewrite the code in the child class

##Abstract Base Class in python

````python
from abc import ABC, abstractmethod


# Yo abstract class ho
class Payment(ABC):

    # Yo payment ko amount store garcha
    def __init__(self, amount):
        self.amount = amount

    # Yo common logging method ho
    def log(self):
        print("Transaction logged")

    # Yo method child class le banaunaiparcha
    @abstractmethod
    def pay(self):
        pass


# Yo eSewa ko payment class ho
class Esewa(Payment):

    # eSewa le payment process garcha
    def pay(self):
        print("eSewa payment:", self.amount)
        self.log()


# Yo Khalti ko payment class ho
class Khalti(Payment):

    # Khalti le payment process garcha
    def pay(self):
        print("Khalti payment:", self.amount)
        self.log()


# Object banaune
e = Esewa(1000)
e.pay()

k = Khalti(500)
k.pay()
````

````python
from abc import ABC, abstractmethod


class PaymentProcessor(ABC):

    def __init__(self, merchant_id):
        self.merchant_id = merchant_id

    # Concrete Method: Common logging shared by all platforms
    def log_transaction(self, amount, status):
        print(
            f"Merchant ID: {self.merchant_id}, "
            f"Amount: ${amount}, Status: {status}"
        )

    # Abstract Method: Each provider handles authorization differently
    @abstractmethod
    def authorize(self, amount):
        pass

    # Abstract Method: Each provider handles payment differently
    @abstractmethod
    def pay(self, amount):
        pass
````

````python
class EsewaProcessor(PaymentProcessor):

    def authorize(self, amount):
        print(f"Authorizing {amount} after eSewa verification")
        return True

    def pay(self, amount):
        if self.authorize(amount):
            print(f"Money {amount} transferred successfully")
            self.log_transaction(amount, "Success")
            return True
        return False
````

````python
class KhaltiProcessor(PaymentProcessor):

    def authorize(self, amount):
        print(f"Authorizing {amount} after Khalti verification")
        return True

    def pay(self, amount):
        if self.authorize(amount):
            print(f"Money {amount} transferred successfully")
            self.log_transaction(amount, "Success")
            return True
        return False
````

````python
#@ Code implementation from the main.py or start.py


def checkout(processor: PaymentProcessor ,total):
  processor.pay(total)
````

###Operator Overloading
- IT allows you to define the behavior of operators for custom objects, you can achieve operator overloading by overriding specific magic method in your class


````python
#@ IMplementation

class Vector:
  def __init__(self, x, y):
    self.x = x
    self.y = y

  def __add__(self, other):
    return Vector(self.x + other.x, self.y + other.y)

  def __sub__(self, other):
    return Vector(self.x - other.x, self.y - other.y)

  def __repr__(self):
    return f"Vector({self.x}, {self.y})"



v1 = Vector(1, 2)
v2 = Vector(3, 4)


v3 = v1 + v2
print(v3)
````
