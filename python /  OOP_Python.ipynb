##Multiple Methods

. A class can have multiple methods that work together

````python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.songs = []          # songs ko empty list

    def add_song(self, song):
        self.songs.append(song)  # song list ma add garxa
        print(f"Added a song: {song}")

    def remove_song(self, song):
        if song in self.songs:
            self.songs.remove(song)  # song remove garxa
            print(f"Removed a song: {song}")

    def show_songs(self):
        print(f"Playlist: {self.name}")  # playlist ko name print garxa
        for song in self.songs:
            print(song)                  # harek song print garxa


my_playlist = Playlist("Favourites")  # object banayo

my_playlist.add_song("Song 1")
my_playlist.add_song("Song 2")
my_playlist.add_song("Song 3")

my_playlist.show_songs()              # songs dekhaune
````

````python
my_playlist.remove_song("Song 3")
````

````python
my_playlist.show_songs()
````

##Single Inheritance

##Multiple Inheritance



````python
class Mother : # parent class 1
  mothername = ""

  def mother(self):
    print(self.mothername)

class Father: # parent class 2
  fathername = ""

  def father(self):
    print(self.fathername)

class Son(Mother, Father): # child class
  def parents(self):
    print("Father :", self.fathername)
    print("Mother :", self.mothername)

s1 = Son()
s1.fathername = "RAM"
s1.mothername = "Sarmila"
s1.parents()
````

## Multilevel Inheritance


````python
#MUltilevel Inheritance grandfather and child

class Grandfather:
  def __init__(self, grandfathername):
    self.grandfathername = grandfathername

class Father(Grandfather):
  def __init__(self, fathername, grandfathername):
    self.fathername = fathername
    Grandfather.__init__(self, grandfathername)

class Son(Father):
  def __init__(self, sonname, fathername, grandfathername):
    self.sonname = sonname
    Father.__init__(self, fathername, grandfathername)

  def print_name(self):
    print('Grandfather name :', self.grandfathername)
    print("Father name :", self.fathername)
    print("Son name :", self.sonname)

s1 = Son('Simon', 'Ram prasad', 'Krishna prasad')
print(s1.grandfathername)
s1.print_name()
````

## Important Notes:

. Son inherits from Father, and Father inherits from GRandFather

. Each constructor passes values up the inheritance chain using explicit constructor calls

. All ancestors class attributes are accessible from the bottom most class which is Son is our case

## Hierarchical INheritance

````python
class Parent:
  def func1(self):
    print("THis function is in parent class")

class Child1(Parent):
  def func2(self):
    print("This function is in child 1")

class Child2(Parent):
  def func3(self):
    print("This function is in child 2")

o1 = Child1()
o2 = Child2()

o1.func1()
o1.func2()

o2.func1()
o2.func3()
````

## Hybrid Inheritance

##Super() function

. super()function that will make the child class inherits all the methods and properties from it's parent class

. It is commonly used in the child class's init() method to initalize inherited attributes.

````python
#@ Animal
#Parent Class
class Animal:
    def __init__(self, name):
        self.name = name

    def info(self):
        print("Animal name:", self.name)


# Child Class
class Dog(Animal):
    def __init__(self, name, breed):
        # Call parent class constructor
        super().__init__(name)
        self.breed = breed

    def details(self):
        print(self.name, "is a", self.breed)


d = Dog("Max", "Husky")

d.info()
d.details()
````

## Takeway:

. ```super()``` function is used ```__init__()``` methof of Dog to call the constructor of ```Animal``` and initialize inherited attribute(name)

. It ensures that parent class functionality is reused without needing to rewrite the code in the child class

````python
#@ Example
class Calculator:
  def multiply(self, a= 5, b=4, *args):
    result = a * b
    for num  in args:
      result *= num

    return result

#creating an object
calc = Calculator()

# using the default params
print(calc.multiply())
````

````python
#@ Overriding
class Animal:
  def sound(self):
    return "Some generic sound"

class Dog(Animal):
  def sound(self):
    return "Woof!"

class Cat(Animal):
  def sound(self):
    return "Meow!"

animals = [Dog(), Cat(), Animal()]
for animal in animals:
  print(animal.sound())
````

Sound method behaves differently depending on whether the object is Dog or Cat or Animal, and this type of decision is made during the runtime.
Thus, this dynamic nature makes Python particularly powerful for runtime Polymorphism,

````python
# Practice Question: Polymorphism with Functions and Methods

# Create a base class Shape with a method area().

# Then create two derived classes:

# - Rectangle — takes width and height and calculates the area.
# - Circle — takes radius and calculates the area.
# Finally, create a function print\_area(shape) that accepts any shape object and prints its area.


# Base class
class Shape:
    def area(self):
        print("Shape ko area")


# Rectangle child class
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    # Rectangle ko area calculate garne method
    def area(self):
        return self.width * self.height


# Circle child class
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    # Circle ko area calculate garne method
    def area(self):
        return 3.14 * self.radius * self.radius


# Jun pani shape object accept garne function
def print_area(shape):
    print("Area:", shape.area())


# Rectangle ko object banaune
rectangle = Rectangle(10, 5)

# Circle ko object banaune
circle = Circle(7)


# Same function le different objects ko area calculate garxa
print_area(rectangle)
print_area(circle)
````
