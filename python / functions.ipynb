```python
def print_numbers(*args):
  for number in args:
    print(number)
```

```python
print_numbers(1,2,3,4,5,6, 'Sejal')
```

```python
def print_numbers(*sejal):
  for number in sejal:
    print(number)
```

```python
print_numbers("Sejal", "Oli")
```

##checking factorial

```python
def factorial(n):
    fact = 1

    for i in range(1, n + 1):
        fact = fact * i

    return fact


num = int(input("Enter a number: "))

print("Factorial:", factorial(num))
```

##Palindrome or not

```python
def check_palindrome(text):
    if text == text[::-1]:
        return "Palindrome"
    else:
        return "Not Palindrome"


word = input("Enter a string: ")

print(check_palindrome(word))
```

##Validate email address

```python
def validate_email(email):
    if "@" in email and "." in email:
        return "Valid Email"
    else:
        return "Invalid Email"


email = input("Enter your email: ")

print(validate_email(email))
```

##Lambda function
  Lambda Function in Python.
  Lambda functions are small anonymonus function defined defined using the keyword Lambda. It can have any numbers of arguments but only one expression. They are commony used for short operations or as arguments to higher order functions

```python
from ast import Expression
##Syntax

lambda arguments: EXpression

<function _main_.<lambda>(arguments)>
```

```python
def add(a, b):
  return a+b
```

```python
add(2,3)
```

```python
add = lambda a, b: a+b
print (add(2,3))
```

```python
check= lambda n: "Even" if n % 2 == 0 else "Odd"

num = int(input("Enter a number: "))

print(check(num))
```
