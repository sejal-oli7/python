```python
import pandas as pd

data = {
    "school": ["s001", "s002", "s003", "s001", "s002", "s004"],
    "age": [12, 12, 13, 13, 14, 12]
}

df = pd.DataFrame(data)

result = df.groupby("school")["age"].agg(
    ["mean", "min", "max"]
)

print(result)
```

```python
import pandas as pd

data = {
    "school": ["s001", "s002", "s003", "s001", "s002", "s004"],
    "class": ["V", "V", "VI", "VI", "V", "VI"],
    "name": [
        "Alberto Franco",
        "Gino Mcneill",
        "Ryan Parkes",
        "Eesha Hinton",
        "Gino Mcneill",
        "David Parkes"
    ],
    "age": [12, 12, 13, 13, 14, 12]
}

df = pd.DataFrame(data)

grouped = df.groupby(["school", "class"])

for name, group in grouped:
    print("Group:", name)
    print(group)
    print()
```
