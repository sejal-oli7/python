##Task 1 leap year

```python
years = int(input("Enter a year :"))
if years %4 == 0:
  if years %100 == 0:
    if years %400 == 0:
      print("It is a leap year")
    else:
      print("It is not a leap year")
  elif years %400 == 0:
    print("It is a leap year")
else:
  print("It is not a leap year")
```

##Task 3 bonus

```python
years = int(input("Enter a year of services :"))
rating = float(input("Enter your rating (1.0 to 5.0):"))
salary = float(input("Enter your salary :"))

bonus_percentage = 0
if rating >= 4.5:
  if years > 10:
    bonus_percentage = 20
  elif years > 5:
    bonus_percentage = 15
  else:
    bonus_percentage = 10

elif rating >= 3.5:
  if years > 10:
    bonus_percentage = 15
  elif years > 5:
    bonus_percentage = 10
  else:
    bonus_percentage = 5

else:
  bonus_percentage = 0

bonus = salary * (bonus_percentage / 100)
print("Bonus Amount: $", round(bonus, 2))
```

##break


```python
# break statement exits the loop prematurely
for i in range (10):
  if i == 5:
    break
  print(i)
```

##continue

```python
# continue statement skips the currrent iteration and continues with others
for i in range (10):
  if i %2 == 0:
    continue
  print(i)
```

##while loop task

```python
n = 10
sum = 0
i = 1
while i <= n:
   sum = sum + i
   i = i+1
print("Sum=", sum)
```

##for loop task

```python
n = 10
sum = 0
for i in range(1, n+1):
   sum = sum + i
print("Sum =", sum)
```
